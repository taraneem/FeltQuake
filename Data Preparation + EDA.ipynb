## Data Understanding

### Dataset Column Descriptions

| Column | Description |
|---|---|
| `mag` | Magnitude of the earthquake |
| `place` | Approximate location description (distance/direction from a known place) |
| `time` | Time the earthquake occurred (UTC) |
| `updated` | Time this record was last updated in USGS's database |
| `tz` | Timezone offset |
| `url` | Link to the USGS event page |
| `detail` | Link to the detailed GeoJSON feed for this specific event |
| `felt` | Number of "Did You Feel It?" reports submitted by the public — **Target Column** |
| `cdi` | Community Determined Intensity — max intensity reported by the public via DYFI |
| `mmi` | Modified Mercalli Intensity — estimated instrumental shaking intensity |
| `alert` | PAGER alert level (green/yellow/orange/red) estimating impact severity |
| `status` | Review status of the event (`automatic` or `reviewed`) |
| `tsunami` | Flag (0/1) indicating tsunami association (mainly oceanic events) |
| `sig` | A significance score combining magnitude, alerts, and felt reports |
| `net` | ID of the network that reported the event (e.g. `us`, `ak`, `pr`) |
| `code` | Identifying code assigned by the contributing network |
| `ids` | List of event IDs associated with this event across networks |
| `sources` | List of contributing networks |
| `types` | List of data product types available for this event |
| `nst` | Number of seismic stations used to locate the earthquake |
| `dmin` | Distance (degrees) from the epicenter to the nearest station |
| `rms` | Root-mean-square residual — indicates location quality/accuracy |
| `gap` | Largest azimuthal gap between stations (degrees) — smaller = more reliable location |
| `magType` | Method used to calculate magnitude (e.g. `ml`, `mb`, `mww`) |
| `type` | Type of seismic event (`earthquake`, `quarry blast`, etc.) |
| `title` | Auto-generated headline combining magnitude and place |
| `longitude` | Epicenter longitude |
| `latitude` | Epicenter latitude |
| `depth` | Depth of the earthquake below the surface (km) |

```python
# Import the required libraries

import numpy as np , pandas as pd , plotly.express as px
import json
```

```python
# Convert GeoJSON to DataFrame

def load_geojson(path):
    with open(path) as f:
        data = json.load(f)
    records = []
    for feature in data["features"]:
        row = dict(feature["properties"])
        row["longitude"], row["latitude"], row["depth"] = feature["geometry"]["coordinates"]
        records.append(row)
    return pd.DataFrame(records)

df1 = load_geojson("Dataset\\query_17_4_2025 to 14_12_2025.json")
df2 = load_geojson("Dataset\\query_16-12_2025 to 18_8_2026.json")

df = pd.concat([df1, df2], ignore_index=True)

df.to_csv("Dataset\\earthquakes_merged.csv", index=False)
print(df.shape)
```

### Why we converted the `time` column?
`time` originally comes as a Unix timestamp (raw milliseconds), which isn't 
usable for analysis on its own. We converted it to a proper datetime so we 
can later engineer features from it (like month, day, or hour) and generate 
insights — for example, plotting how earthquake frequency changes month by 
month across 2025.


```python
df['time'] = pd.to_datetime(df['time'], unit='ms')
df['time']
```

### Handling missing [administrative columns]
`tz` is 100% empty and provides no information, so it was dropped along 
with other purely administrative columns (`url`, `detail`, `title`, `ids`, 
`sources`, `types`, `code`, `updated`) that don't represent physical 
earthquake properties (metadata about the record itself).
They can't help the model learn anything meaningful

```python
dropped_columns = ['tz', 'url', 'detail', 'title', 'ids', 'sources', 'types', 'code', 'updated']

df.drop(columns=dropped_columns, inplace=True)
```

```python
df.duplicated().sum()
```

`cdi`, `mmi`, and `alert` are kept for now to visualize their relationship 
with `felt` before excluding them from training due to data leakage. 

`alert` is heavily missing (96.7%) and will also be excluded — not for 
leakage, but because too little real data exists to use it reliably.

`nst`, `dmin`, `rms`, `gap` drop NANS as it's a small percentage.


```python
drop_nans = ['nst' , 'dmin' , 'rms' , 'gap']

df.dropna(subset=drop_nans, inplace=True , ignore_index=True)
```

### Why `cdi` and `mmi` are excluded from the model?
A crosstab of `felt.notna()` vs `cdi.notna()` showed a perfect 100% match: 
every row with `felt` also has `cdi`, and vice versa, with zero exceptions 
— meaning `cdi`'s mere presence directly reveals the target. This is 
confirmed data leakage.

```python
print(pd.crosstab(df['felt'].notna(), df['cdi'].notna()))
```

`mmi` showed a weaker pattern: it doesn't align perfectly with `felt` 
(some quakes have `mmi` without `felt`, and vice versa), so it isn't a 
guaranteed leak. but also it's better to exclude it from data. 

```python
print(pd.crosstab(df['felt'].notna(), df['mmi'].notna()))
```

```python
excluded_cols =['mmi', 'cdi']
df.drop(columns= excluded_cols , inplace= True)
```

- Handling Target Column (`felt`)

`felt` originally holds the number of "Did You Feel It?" reports, with 
missing values (`NaN`) where no one reported feeling the earthquake. We 
converted it into a binary target, `reported`:
- `NaN` or `0` → `0` (not reported / not felt)
- Any value greater than `0` → `1` (reported / felt)

```python
df['reported'] = (df['felt'] > 0).astype(int)
```

```python
df['reported'].isna().sum()
```

```python
df.drop('felt' , axis=1 , inplace= True)
```

## Feature Engineering

**Time-based features** — extracted from `time` since the raw datetime 
isn't directly usable by a model:
- `hour` — hour of day (0–23)
- `month` — month (1–12)
- `day_of_week` — day of week (0–6)

```python
df['hour'] = df['time'].dt.hour
df['month'] = df['time'].dt.month
df['day_of_week'] = df['time'].dt.dayofweek
df.drop('time' , axis=1 , inplace=True)
```

`mag_depth_ratio` vs. `sig`

**Why it matters:** a magnitude-5 earthquake at 5 km depth is far more 
likely to be felt than a magnitude-5 earthquake at 500 km depth, even 
though they share the same magnitude. `mag` alone doesn't capture this 
difference, but `mag_depth_ratio` summarizes it in one number — a higher 
ratio means shallower and stronger, and therefore more likely to be felt.

`sig` is a ready-made score calculated by USGS using an internal formula 
that factors in magnitude, alerts, and actual public "felt" reports — so 
it's partly built from human reaction data.

```python
df['mag_depth_ratio'] = df['mag'] / (df['depth'] + 1)
```

`place` is free text and not directly usable, so we extract one signal: 
does it contain "km" (near a populated place) or not (remote region).


```python
df['is_remote'] = ~df['place'].str.contains('km', case=False, na=False)
df.drop('place', axis=1 , inplace=True) 
```

- Does the time of day or month affect reporting?

```python
# Convert 'hour' into meaningful values due to the analysis view.

def hour_to_label(h):
    if h == 0:
        return "12 AM"
    elif h < 12:
        return f"{h} AM"
    elif h == 12:
        return "12 PM"
    else:
        return f"{h - 12} PM"

df['hour_label'] = df['hour'].apply(hour_to_label)
```

```python
hour_order = [hour_to_label(h) for h in range(24)] # sorting 

plot_df = df.groupby('hour_label')['reported'].sum().reset_index()
plot_df
```

```python
px.bar(
    plot_df,
    x='hour_label',
    y='reported',
    title='Reporting Rate by Hour of Day',
    color_discrete_sequence=["#3B5E64"],
    labels={'hour_label': 'Hour of Day', 'reported': 'Reporting Rate'},
    text_auto=True,        
    category_orders={'hour_label': hour_order ,
                     }  
).update_xaxes(categoryorder='max descending')
```

```python
month_map = {
    1: 'January', 2: 'February', 3: 'March', 4: 'April',
    5: 'May', 6: 'June', 7: 'July', 8: 'August',
    9: 'September', 10: 'October', 11: 'November', 12: 'December'
}

df['month_name'] = df['month'].map(month_map)
```

```python
plot_df = df.groupby('month_name')['reported'].mean() .reset_index(name='Average Reporting Rate').sort_values(ascending=False , by='Average Reporting Rate')
plot_df['Average Reporting Rate'] = round(plot_df['Average Reporting Rate'] * 100 , 2)
plot_df
```

```python
# month_order = list(month_map.values())

px.line(
    plot_df,
    x='month_name',
    y='Average Reporting Rate',
    title='Average Reporting Rate by Month',
    markers=True,                          
    labels={'month_name': 'Month'},
    # category_orders={'month_name': month_order}
)
```
